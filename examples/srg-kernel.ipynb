# Static Rotational Graph (SRG) kernel 

The SRG kernel is designed to provide an extensible tiling, where neighbour/adjacency relations are generated during construction, rendering a subsequent computation of cell neighbours obsolete

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from hypertiling import HyperbolicTiling
from hypertiling.graphics.svg import SVGCanvas, Tiling, UnitCircle, draw_svg
from hypertiling.graphics.plot import plot_tiling
```

```python
p, q, n = 3, 7, 2
T = HyperbolicTiling(p, q, n, kernel="SRG", center="vertex")
T.add_layer()
```

Access neighbours

```python
print(T.get_nbrs(1))
print(T.get_nbrs(9))
```

New layers can be added anytime

```python
T.add_layer()
```

```python
# Calculate distances for coloring
# We compute the Euclidean distance of each cell's center (T[i][0]) from the origin (0,0)
# Note: T[i][0] returns the [x, y] coordinates of the center.
dists = [np.linalg.norm(T[i][0]) for i in range(len(T))]

# Map distances to colors
# We normalize the distances and apply a colormap (e.g., "magma_r" or "viridis")
norm = plt.Normalize(min(dists), max(dists))
cmap = plt.get_cmap("YlOrBr") # 'magma_r' is often nice for hyperbolic disks
facecolors = [mcolors.to_hex(cmap(norm(d))) for d in dists]

# 4. Compose the SVG and render
canvas = SVGCanvas()
canvas.add(UnitCircle())
canvas.add(Tiling(T, facecolors=facecolors, edgecolor="black", lw=0.5))
svg_content = canvas.render()
draw_svg(svg_content)

# Save to file
# Use standard Python file writing
with open("dummy.svg", "w") as f:
    f.write(svg_content)
```

```python
plot_tiling(T, dpi=220)
for i in range(len(T)):
    z = T.get_center(i)
    l = T.get_layer(i)
    t = plt.text(z.real, z.imag, "", fontsize=15-4*l, ha="center", va="center")
```

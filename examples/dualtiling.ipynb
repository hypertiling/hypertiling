```python
from hypertiling import GraphKernels, HyperbolicGraph
from hypertiling import TilingKernels, HyperbolicTiling
from hypertiling.graphics.plot import plot_tiling, quick_plot, plot_geodesic, convert_edges_to_arcs
import hypertiling.graphics.svg as svg
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Arc
from hypertiling.geodesics import geodesic_arc
```

```python
import cmath

def hyperbolic_triangle_center_approx(z1: complex, z2: complex, z3: complex) -> complex:
    """
    Approximates the center of a triangle in the Poincaré disk model.
    This is a simple approximation by taking the Euclidean average of the
    complex coordinates and then normalizing to ensure it's within the disk.

    Args:
        z1: Complex coordinate of the first vertex.
        z2: Complex coordinate of the second vertex.
        z3: Complex coordinate of the third vertex.

    Returns:
        A complex number representing the approximate center of the triangle.
    """
    # Calculate the Euclidean average of the complex coordinates
    euclidean_centroid = (z1 + z2 + z3) / 3

    # Normalize the result to ensure it's within the unit disk.
    # If the magnitude is > 1, scale it back to be within the disk.
    # This is a very simple projection.
    magnitude = abs(euclidean_centroid)
    if magnitude > 1.0:
        return euclidean_centroid / magnitude
    else:
        return euclidean_centroid
```

```python
color_graph = "#A2CE3CFF"
color_tiling = "#000000"


# plot tiling and dual tiling (=graph)
def plot_both(g, fname):

    plt.figure()
    ax = plt.gca()
    
    # ----- draw tiling -----
    for edge in convert_edges_to_arcs(g)[0]:
        ax.add_artist(edge)
        edge.set_color(color_tiling)
        edge.set_lw(0.5)


    # ----- draw dual tiling -----
    edges = []
    for i in range(len(g)):
        center1 = hyperbolic_triangle_center_approx(*g.get_vertices(i)[1:])

        nbrs = g.get_nbrs(i)

        for j in range(len(nbrs)):
            center2 = hyperbolic_triangle_center_approx(*g.get_vertices(nbrs[j])[1:])
            edges += [geodesic_arc(center1, center2)]
        
    for tedge in edges:
        ax.add_artist(tedge)
        tedge.set_color(color_graph)
        tedge.set_lw(0.8)
        tedge.set_zorder(5)
        
    
    plt.axis('off')
    plt.xlim(-1,1)
    plt.ylim(-1,1)
    plt.gca().set_aspect('equal')
    plt.savefig(fname, dpi=500, bbox_inches='tight')
    plt.show()
```

```python
g = HyperbolicTiling(5, 4, 2, 20, kernel=TilingKernels.GRCT, nbrs=True)

plot_both(g, "tiling_5_4_2_20.pdf")
```

```python
g = HyperbolicTiling(2, 3, 7, 22, kernel=TilingKernels.GRCT, nbrs=True)

plot_both(g, "tiling_2_3_7_22.pdf")
```
